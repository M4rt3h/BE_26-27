```python
 # function to solve forward kinematics
def solveForwardPositionKinematics(robot_id, joint):
    print('Forward position kinematics')
    # get end-effector link state
    eeState = p.getLinkState(robot_id, joint)
    link_trn, link_rot, com_trn, com_rot, frame_pos, frame_rot = eeState
#    eePose = list(link_trn) + list(p.getEulerFromQuaternion(link_rot))
    eePose = list(link_trn) + list(link_rot)
    print('End-effector pose:', eePose)
    return eePose
```

```python
but, dbut, ddbut=[],[],[]
xbut=[]
x= list(position['x'])
y= list(position['y'])
z= list(position['z'])
for i,j,k in zip(x,y,z):
    but.append(np.linalg.norm([i,j,k]))
    xbut.append([i,j,k])
    
dx= list(vitesse['x'])
dy= list(vitesse['y'])
dz= list(vitesse['z'])
for i,j,k in zip(dx,dy,dz):
    dbut.append(np.linalg.norm([i,j,k]))

vitesseOT= np.linalg.norm(dbut)
print("VitesseOT",dbut, "  et   ", vitesseOT)

affiche.affiche_courbe2D(10,'de vitesse OT (m/s)',temps,dbut,'r')

ddx= list(acceleration['x'])
ddy= list(acceleration['y'])
ddz= list(acceleration['z'])
for i,j,k in zip(ddx,ddy,ddz):
    ddbut.append(np.linalg.norm([i,j,k]))
affiche.affiche_courbe2D(10,'accélération OT',temps,ddbut,'r')
#print("But=",but)
```

```python
ori= [0,0,0,1]

for pos in xbut:
    qinv = list(p.calculateInverseKinematics(robot, 6,pos, ori))
    print("Pos=",pos)
    print(" qi sol du MGI pour UR3=", qinv, "type=",type(qinv))
    control.PositionAffiche(robot, control_joints, qinv, frame_ee)
    q,dq,torq= control.getJointStates(robot, control_joints)

# Récupérer la vitesse linéaire (v_lin) et angulaire (v_ang)
objets_dynamiques = [cube_id]
# Seuil en dessous duquel on considère que l'objet est immobile (en m/s et rad/s)
SEUIL_STATIQUE = 0.001 

print("Simulation en cours... En attente de la stabilisation des objets.")

simulation_active = True
while simulation_active:
    p.stepSimulation()
    time.sleep(1./240.)
    
    tous_immobiles = True
    time.sleep(0.1)
    for obj_id in objets_dynamiques:
        # Récupérer la vitesse linéaire (v_lin) et angulaire (v_ang)
        v_lin, v_ang = p.getBaseVelocity(obj_id)
        
        # Calculer la norme (la vitesse globale)
        vitesse_lineaire_norme = np.linalg.norm(v_lin)
        vitesse_angulaire_norme = np.linalg.norm(v_ang)
        print("seuil stat=", vitesse_lineaire_norme, " et  ",vitesse_angulaire_norme)
        p.stepSimulation()
        # Si un seul objet dépasse le seuil, le système n'est pas statique
        if vitesse_lineaire_norme > SEUIL_STATIQUE or vitesse_angulaire_norme > SEUIL_STATIQUE:
            tous_immobiles = False
            break # Pas la peine de vérifier les autres objets pour cette frame
            
    # Si après avoir vérifié tous les objets, ils sont tous sous le seuil : on arrête
    if tous_immobiles:
        print("Tous les objets sont statiques ! Fin de la simulation.")
        simulation_active = False

# Optionnel : Laisser la fenêtre ouverte sans simuler la physique pour observer le résultat
print("Affichage figé. Fermez la fenêtre manuellement.")
while p.isConnected():
    time.sleep(0.1)
```
